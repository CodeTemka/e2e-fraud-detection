```python
import os
import sys

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '../..')))

from fraud_detection.azure.client import get_ml_client
ml_client = get_ml_client()
from azure.ai.ml import command, Input, Output, dsl
```

```python
import uuid

job_name = f"custom_lightgbm_training_job_{uuid.uuid4().hex[:6]}"

job = command(
    name=job_name,
    description="Train a LightGBM model for fraud detection with MLflow logging and model registration",
    inputs={
        "data_path": Input(
            type="uri_file",
            path=ml_client.data.get(name="creditcard-data", version='1').path
        )
    },
    code=".",  # project root
    command="python train_script.py --data_path ${{inputs.data_path}}",
    environment="azureml://registries/azureml/environments/lightgbm-3.3/versions/69",
    compute="automated-ml-cluster",
    experiment_name="custom_train_experiment",
    display_name="Custom LightGBM Training Job"
)
```

```python
returned_job = ml_client.jobs.create_or_update(job)
ml_client.jobs.stream(returned_job.name)
```
